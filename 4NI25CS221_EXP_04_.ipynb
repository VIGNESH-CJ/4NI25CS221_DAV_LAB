```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 rows of data")
print(df.head())
```

```python
print("\nDataset Info")
print(df.info())
```

```python
print("\nMissing values before handling : ")
print(df.isnull().sum())
```

```python
df.loc[0:5,"Study_Hours_per_Week"] = None
df.loc[10:15,"Attendance_Rate"] = None
df.loc[20:25,"Past_Exam_Scores"] = None
df.loc[30:35,"Final_Exam_Score"] = None

print("Missing values after handling : ")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

Mean Imputation

```python
df["Study_Hours_per_Week"] = df["Study_Hours_per_Week"].fillna(df["Study_Hours_per_Week"].mean())
```

Median Imputation

```python
df["Attendance_Rate"] = df["Attendance_Rate"].fillna(df["Attendance_Rate"].median())
```

Forward Imputation

```python
df["Past_Exam_Scores"] = df["Past_Exam_Scores"].ffill()
```

```python
df["Final_Exam_Score"] = df["Final_Exam_Score"].fillna(df["Final_Exam_Score"].mean())
print("\nMissing values before Inputation : ")
print(df.isnull().sum())
```

One-Hot_Encoding

```python
df = pd.get_dummies(df,columns=['Parental_Education_Level_Masters'])
print("\One_Hot Encoded Columns : ")
print(df.columns.tolist())
```

```python
le=LabelEncoder()
df['Pass_Fail_Encoded']=le.fit_transform(df['Pass_Fail'])
print("\n Label Encoded Result")
print(df[["Pass_Fail","Pass_Fail_Encoded"]].head())
```

```python
df["Gender_Encoded"] = le.fit_transform(df["Gender"])
df["Internet_Access_Encoded"] = le.fit_transform(df["Internet_Access_at_Home"])
df["Extracurricular_Activities_Encoded"] = le.fit_transform(df["Extracurricular_Activities"])

```

```python
df.to_csv("student_performance_Cleaned_dataset.csv", index=False)
print("\n Cleand dataset saved as 'student_performance_dataset.csv'")
```
