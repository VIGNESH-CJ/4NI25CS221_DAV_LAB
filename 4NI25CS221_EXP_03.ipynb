```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/student_data.csv")
print("Dataset shape:  ",df.shape)
```

**Histograms: Display the distribution of numerical data by grouping values into bins, helping identify patterns or frequency **

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"], bins=20, kde=True, color="steelblue")
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x="studytime",y="G3",data=df, palette="Set2")
plt.title("Final Grade across Study Time Levels")
plt.xlabel("Study Time (  1-Low  ||  4-High  )")
plt.ylabel("Final Grade")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x="G1", y="G3" ,data=df, hue="sex", palette="coolwarm")
plt.title("Final Period Grade vs Final Grade ")
plt.xlabel("Period 1 Grade (G1)")
plt.ylabel(" Final Grade (G3)")
plt.show()
```

**4. PAIRPLOT - pairwise relations across key numeric columns**


```python
sns.pairplot(df[["age", "studytime","G1", "G2" ,"G3","sex"]], hue="sex", palette="coolwarm")
plt.suptitle("Pairwise Relationship", y=1.02)
plt.show()
```

**5. Correlation Heatmap - Strength of relationships between numeric features**

```python
numeric_columns=["age", "studytime","G1", "G2" ,"G3","failures","absences"]
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_columns].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
