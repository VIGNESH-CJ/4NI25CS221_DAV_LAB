```python
data={
    "Name":["AJ","BJ","CJ","DJ","EG"],
    "Marks":[98,89,99,85,86],
    "Age":[19,20,21,18,19],
    "Branch":["CSE","ECE","CSE","ISE","ME"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print("\n slicing: \n",df.iloc[0:3])
print("Marks: \n",df['Marks'])
```

```python
print("Marks > 90 : \n",df[df["Marks"]>90])
```
